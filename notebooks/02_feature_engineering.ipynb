## Feature dictionary

| Feature | Definition |
|---|---|
| Age | 2014 − Year_Birth (data ends June 2014) |
| Children | Kidhome + Teenhome |
| Family_Size | 1 + partner (Married/Together) + Children |
| Education_Level | Basic/2n Cycle = 0, Graduation = 1, Master/PhD = 2 |
| Total_Spent | Sum of the six Mnt* columns |
| Total_Purchases | Web + Catalogue + Store purchases |
| Total_Accepted | AcceptedCmp1–5 + Response |
| Customer_Days | Days between enrolment and the latest enrolment date |

```python
fig, axes = plt.subplots(2, 4, figsize=(16, 7))
for ax, col in zip(axes.flat, ["Age", "Children", "Family_Size", "Education_Level", "Total_Spent", "Total_Purchases", "Total_Accepted", "Customer_Days"]):
    sns.histplot(df[col], ax=ax, bins=20)
    ax.set_title(col)
plt.tight_layout(); plt.show()
```

```python
plt.figure(figsize=(12, 9))
sns.heatmap(df[FEATURES].corr(), cmap="RdBu_r", center=0, annot=True, fmt=".1f", annot_kws={"size": 7})
plt.title("Correlation of clustering features"); plt.show()
```
