```python
import sys, warnings
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))   # so `src` can be imported from notebooks/
warnings.filterwarnings("ignore")

import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", 50)
FIG = Path.cwd().parent / "reports" / "figures"
```

```python
from src.preprocess import load_clean_data, MNT_COLS, PURCHASE_COLS, CAMPAIGN_COLS

df = load_clean_data()
df.shape
```

## Distributions

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, ["Age", "Income", "Total_Spent"]):
    sns.histplot(df[col], kde=True, ax=ax); ax.set_title(col)
plt.tight_layout(); plt.savefig(FIG / "distributions.png", dpi=150); plt.show()
```

## Income vs spending

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Income", y="Total_Spent", hue="Children", palette="viridis", alpha=0.6)
plt.title("Higher income → higher spend; children lower spend")
plt.savefig(FIG / "income_vs_spend.png", dpi=150); plt.show()
```

## Spending by children and education

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.boxplot(data=df, x="Children", y="Total_Spent", ax=axes[0])
sns.boxplot(data=df, x="Education", y="Total_Spent", order=["Basic", "2n Cycle", "Graduation", "Master", "PhD"], ax=axes[1])
plt.tight_layout(); plt.show()
```

## Product categories and purchase channels

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
df[MNT_COLS].sum().sort_values().plot.barh(ax=axes[0], title="Total spend by product")
df[PURCHASE_COLS].sum().sort_values().plot.barh(ax=axes[1], title="Purchases by channel")
plt.tight_layout(); plt.show()
```

## Campaign acceptance

```python
(df[CAMPAIGN_COLS].mean() * 100).plot.bar(title="Acceptance rate per campaign (%)")
plt.show()
```
